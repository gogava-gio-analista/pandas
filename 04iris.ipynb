```python
df_grupo = df.groupby('class')['sepal_longitud'].mean()
df_grupo.plot(kind='bar')
plt.ylabel('Media de sepal_longitud')
plt.title('Media de sepal_longitud por clase')
ax = plt.gca()
ax.bar_label(ax.containers[0], fmt='%.2f') 
plt.show()
```
