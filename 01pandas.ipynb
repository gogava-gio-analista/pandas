```python
df_datos = pd.read_csv('data/datos.csv', delimiter=';')
df_datos
```

```python
from matplotlib.ticker import MaxNLocator

df_agrupado = df_datos.groupby('ocupacion')['ocupacion'].count().astype(int)
df_agrupado.plot(kind='bar')
plt.xlabel('Ocupación')
plt.ylabel('Cantidad')
plt.title('Cantidad de personas por ocupación')
plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
plt.xticks(rotation=45)
plt.show()
```

```python
# trazo una línea con los datos del csv (ya agrupados en df_agrupado)
# uso un colormap para que cada punto/segmento tenga un color distinto
colors = plt.cm.tab10(range(len(df_agrupado)))

df_agrupado.plot(kind='line', color=colors, marker='o')
plt.xlabel('Ocupación')
plt.ylabel('Cantidad')
plt.title('Cantidad de personas por ocupación (línea)')
plt.xticks(rotation=45)
plt.show()
```
