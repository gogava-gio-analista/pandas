```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('data/game_of_thrones.csv')
df.head(2)
```

```python
df.shape
```

```python
df.info()
```

```python
df.fillna({'allegiance': 'Unknown', 'killer': 'Unknown', 'killers_house': 'Unknown'}, inplace=True)
df.info()
```

```python
df_agrupado_temporada = df.groupby('season').size()
ax = df_agrupado_temporada.plot(kind='bar')
plt.title('Número de Muertes por Temporada en Game of Thrones')
plt.xlabel('Temporada')
plt.ylabel('Número de Muertes')
plt.xticks(rotation=0)

# calcular porcentajes y mostrarlos arriba de cada barra
total = df_agrupado_temporada.sum()
for p in ax.patches:
    height = p.get_height()
    pct = height / total * 100
    ax.annotate(f"{pct:.1f}%", (p.get_x() + p.get_width() / 2, height),
                ha='center', va='bottom')

plt.show()
```
