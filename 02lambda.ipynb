```python
listanumerica = [1, 2, 3, 4, 5, 23, 45, 67, 89, 100]
```

Vamos a multiplicar cada uno en 2 y ver su doble 

```python
listadoble = list(map(lambda x: x * 2, listanumerica))
listadoble
```

```python
tuplanumerica = (1, 2, 3, 4, 5, 23, 45, 67, 89, 100)
tupladoble = tuple(map(lambda x: x * 2, tuplanumerica))
tupladoble
```

```python
dictnumerica = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
dictdoble = dict(map(lambda item: (item[0], item[1] * 2), dictnumerica.items()))
dictdoble
```

```python
listanumericaconnegativo = [1, 2, 3, -4, 5, 23, -45, 67, 89, -100]
listadoblepositivo = list(filter(lambda x: x > 0, listanumericaconnegativo))
listadoblepositivo
```
